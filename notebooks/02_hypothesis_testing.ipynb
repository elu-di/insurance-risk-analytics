
# Hypothesis Testing for Risk Drivers
This notebook will statistically validate or reject key hypotheses about risk drivers.


```

import sys
sys.path.append('../src')
import pandas as pd
import numpy as np
from hypothesis_tests import chi_squared_test, t_test_independent, calculate_margin, define_claim_frequency

# Load data
data = pd.read_csv('../data/insurance_data.csv')
data = calculate_margin(data)
data = define_claim_frequency(data)

# Print basic stats
print("Total rows:", len(data))
```

```

# H0: There are no risk differences across provinces.
# Risk metric = Claim_Occurred (Categorical)
p_val_province = chi_squared_test(data, 'Province', 'Claim_Occurred')
print("P-value for Province vs Claim_Occurred:", p_val_province)
decision = 'Reject H0' if p_val_province < 0.05 else 'Fail to reject H0'
print("Decision:", decision)
```

```

# H0: There are no risk differences between zip codes.
# Due to many zip codes, let's select two zip codes that are most common
top_zips = data['ZipCode'].value_counts().head(2).index.tolist()
zip_data = data[data['ZipCode'].isin(top_zips)]
p_val_zip_risk = chi_squared_test(zip_data, 'ZipCode', 'Claim_Occurred')
print(f"P-value for Zip Codes {top_zips} vs Claim_Occurred:", p_val_zip_risk)
decision = 'Reject H0' if p_val_zip_risk < 0.05 else 'Fail to reject H0'
print("Decision:", decision)
```

```

# H0: There is no significant margin (profit) difference between zip codes.
margin_zip1 = zip_data[zip_data['ZipCode'] == top_zips[0]]['Margin']
margin_zip2 = zip_data[zip_data['ZipCode'] == top_zips[1]]['Margin']
p_val_zip_margin = t_test_independent(margin_zip1, margin_zip2)
print(f"P-value for Zip Codes {top_zips} Margin difference:", p_val_zip_margin)
decision = 'Reject H0' if p_val_zip_margin < 0.05 else 'Fail to reject H0'
print("Decision:", decision)
```

```

# H0: There is no significant risk difference between Women and Men.
gender_data = data[data['Gender'].isin(['Male', 'Female', 'M', 'F'])]
# standardizing gender
gender_data['Gender'] = gender_data['Gender'].replace({'M': 'Male', 'F': 'Female'})
p_val_gender = chi_squared_test(gender_data, 'Gender', 'Claim_Occurred')
print("P-value for Gender vs Claim_Occurred:", p_val_gender)
decision = 'Reject H0' if p_val_gender < 0.05 else 'Fail to reject H0'
print("Decision:", decision)
```
